```python
# -*- coding: utf-8 -*-
# PyTorch multi-layer, multi-neuron model (without OOP)

import torch
import torch.nn.functional as F

# ==============================================
# 1️⃣ Weight and bias definitions (manual)
# ==============================================
# Layer dimensions:
# 1 → 8 → 6 → 4 → 1
w1 = torch.randn(1, 8, requires_grad=True)
b1 = torch.zeros(8, requires_grad=True)

w2 = torch.randn(8, 6, requires_grad=True)
b2 = torch.zeros(6, requires_grad=True)

w3 = torch.randn(6, 4, requires_grad=True)
b3 = torch.zeros(4, requires_grad=True)

w4 = torch.randn(4, 1, requires_grad=True)
b4 = torch.zeros(1, requires_grad=True)

# ==============================================
# 2️⃣ Training data (example input-output)
# ==============================================
x = torch.linspace(-3, 3, 100).reshape(-1, 1)
y = 2.0 * torch.cos(x) + 0.3 * torch.exp(torch.sin(x))  # target function

# ==============================================
# 3️⃣ Learning parameters
# ==============================================
lr = 0.01
optimizer = torch.optim.Adam([w1, b1, w2, b2, w3, b3, w4, b4], lr=lr)
loss_fn = torch.nn.MSELoss()

# ==============================================
# 4️⃣ Training loop (manual forward & backward)
# ==============================================
for epoch in range(2000):
    # --------- FORWARD PASS ---------
    # 1st layer (1→8, sine activation)
    z1 = x @ w1 + b1           # (100×1) @ (1×8) = (100×8)
    a1 = torch.sin(z1)         # sine activation

    # 2nd layer (8→6, exponential activation)
    z2 = a1 @ w2 + b2          # (100×8) @ (8×6) = (100×6)
    a2 = torch.exp(torch.clamp(z2, max=4))  # exponential activation (overflow protected)

    # 3rd layer (6→4, cosine activation)
    z3 = a2 @ w3 + b3          # (100×6) @ (6×4) = (100×4)
    a3 = torch.cos(z3)

    # 4th output layer (4→1, linear)
    y_pred = a3 @ w4 + b4      # (100×4) @ (4×1) = (100×1)

    # --------- LOSS & BACKPROP ---------
    loss = loss_fn(y_pred, y)
    optimizer.zero_grad()      # reset gradients
    loss.backward()            # automatic backpropagation
    optimizer.step()           # update weights

    # --------- MONITORING ---------
    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | Loss = {loss.item():.6f}")

# ==============================================
# 5️⃣ Post-training results
# ==============================================
with torch.no_grad():
    sample_x = torch.tensor([[0.5]])
    pred = sample_x @ w1 + b1
    pred = torch.sin(pred)
    pred = pred @ w2 + b2
    pred = torch.exp(torch.clamp(pred, max=4))
    pred = pred @ w3 + b3
    pred = torch.cos(pred)
    pred = pred @ w4 + b4
    print("\nSample prediction (x=0.5):", pred.item())
    print("Final Loss:", loss.item())
```
