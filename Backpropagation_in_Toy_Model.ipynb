```python
# -*- coding: utf-8 -*-
"""7_1_Backpropagation_in_Toy_Model.ipynb
Automatically generated by Colab.
"""

# ==========================================================
#  IMPORT LIBRARIES
# ==========================================================
import numpy as np

# ==========================================================
#  ORIGINAL FUNCTION AND LOSS
# ==========================================================
def fn(x, beta0, beta1, beta2, beta3, omega0, omega1, omega2, omega3):
    return beta3 + omega3 * np.cos(beta2 + omega2 * np.exp(beta1 + omega1 * np.sin(beta0 + omega0 * x)))

def loss(x, y, beta0, beta1, beta2, beta3, omega0, omega1, omega2, omega3):
    diff = fn(x, beta0, beta1, beta2, beta3, omega0, omega1, omega2, omega3) - y
    return diff * diff

# ==========================================================
#  INITIAL VALUES
# ==========================================================
beta0 = 1.0; beta1 = 2.0; beta2 = -3.0; beta3 = 0.4
omega0 = 0.1; omega1 = -0.4; omega2 = 2.0; omega3 = 3.0
x = 2.3; y = 2.0

l_i_func = loss(x, y, beta0, beta1, beta2, beta3, omega0, omega1, omega2, omega3)
print('l_i = %3.3f' % l_i_func)

# ==========================================================
#  MANUAL DERIVATIVE (dL/dω0)
# ==========================================================
dldomega0_func = -2 * (beta3 + omega3 * np.cos(beta2 + omega2 * np.exp(beta1 + omega1 * np.sin(beta0 + omega0 * x))) - y) * \
              omega1 * omega2 * omega3 * x * np.cos(beta0 + omega0 * x) * \
              np.exp(beta1 + omega1 * np.sin(beta0 + omega0 * x)) * \
              np.sin(beta2 + omega2 * np.exp(beta1 + omega1 * np.sin(beta0 + omega0 * x)))

# Finite difference check
dldomega0_fd = (loss(x, y, beta0, beta1, beta2, beta3, omega0 + 0.00001, omega1, omega2, omega3)
                - loss(x, y, beta0, beta1, beta2, beta3, omega0, omega1, omega2, omega3)) / 0.00001

print('dL/dω0: Function value = %3.3f, Finite difference value = %3.3f' % (dldomega0_func, dldomega0_fd))

# ==========================================================
#  FORWARD PASS
# ==========================================================
f0 = beta0 + omega0 * x
h1 = np.sin(f0)
f1 = beta1 + omega1 * h1
h2 = np.exp(f1)
f2 = beta2 + omega2 * h2
h3 = np.cos(f2)
f3 = beta3 + omega3 * h3
l_i = (f3 - y) ** 2

print("\n--- Forward Pass Check ---")
print("f0: true = 1.230, yours = %3.3f" % f0)
print("h1: true = 0.942, yours = %3.3f" % h1)
print("f1: true = 1.623, yours = %3.3f" % f1)
print("h2: true = 5.068, yours = %3.3f" % h2)
print("f2: true = 7.137, yours = %3.3f" % f2)
print("h3: true = 0.657, yours = %3.3f" % h3)
print("f3: true = 2.372, yours = %3.3f" % f3)
print("l_i: original = %3.3f, from forward pass = %3.3f" % (l_i_func, l_i))

# ==========================================================
#  BACKWARD PASS
# ==========================================================
# Step-by-step derivatives
dldf3 = 2 * (f3 - y)
dldh3 = omega3 * dldf3
dldf2 = (-np.sin(f2)) * dldh3          # dh3/df2 = -sin(f2)
dldh2 = omega2 * dldf2                 # df2/dh2 = omega2
dldf1 = np.exp(f1) * dldh2             # dh2/df1 = exp(f1)
dldh1 = omega1 * dldf1                 # df1/dh1 = omega1
dldf0 = np.cos(f0) * dldh1             # dh1/df0 = cos(f0)

print("\n--- Backward Pass Check ---")
print("dldf3: true = 0.745, yours = %3.3f" % dldf3)
print("dldh3: true = 2.234, yours = %3.3f" % dldh3)
print("dldf2: true = -1.683, yours = %3.3f" % dldf2)
print("dldh2: true = -3.366, yours = %3.3f" % dldh2)
print("dldf1: true = -17.060, yours = %3.3f" % dldf1)
print("dldh1: true = 6.824, yours = %3.3f" % dldh1)
print("dldf0: true = 2.281, yours = %3.3f" % dldf0)

# ==========================================================
#  FINAL PARAMETER DERIVATIVES
# ==========================================================
dldbeta3 = dldf3
dldomega3 = dldf3 * h3
dldbeta2 = dldf2
dldomega2 = dldf2 * h2
dldbeta1 = dldf1
dldomega1 = dldf1 * h1
dldbeta0 = dldf0
dldomega0 = dldf0 * x

print("\n--- Final Parameter Gradients ---")
print('dldbeta3 : your = %3.3f | true = 0.745' % dldbeta3)
print('dldomega3: your = %3.3f | true = 0.489' % dldomega3)
print('dldbeta2 : your = %3.3f | true = -1.683' % dldbeta2)
print('dldomega2: your = %3.3f | true = -8.530' % dldomega2)
print('dldbeta1 : your = %3.3f | true = -17.060' % dldbeta1)
print('dldomega1: your = %3.3f | true = -16.079' % dldomega1)
print('dldbeta0 : your = %3.3f | true = 2.281' % dldbeta0)
print('dldomega0: your = %3.3f | func = %3.3f | finite diff = %3.3f'
      % (dldomega0, dldomega0_func, dldomega0_fd))
```
