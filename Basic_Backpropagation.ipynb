```python
import numpy as np

# -----------------------------
# 1. Activation functions
# -----------------------------
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)  # x here is the output of sigmoid(x)

# -----------------------------
# 2. Training data (X, y)
# -----------------------------
# Input - 4 samples, 2 features
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

# Target output - XOR problem
y = np.array([[0],
              [1],
              [1],
              [0]])

# -----------------------------
# 3. Random initialization of weights
# -----------------------------
np.random.seed(42)
input_size = 2
hidden_size = 2
output_size = 1

W1 = np.random.rand(input_size, hidden_size) - 0.5
W2 = np.random.rand(hidden_size, output_size) - 0.5

# -----------------------------
# 4. Learning rate
# -----------------------------
lr = 0.5

# -----------------------------
# 5. Training loop
# -----------------------------
for epoch in range(10000):

    # ---- FORWARD PASS ----
    hidden_input = np.dot(X, W1)
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2)
    final_output = sigmoid(final_input)

    # ---- ERROR (LOSS) ----
    error = y - final_output

    # ---- BACKWARD PASS ----
    # Gradient at the output layer
    d_output = error * sigmoid_derivative(final_output)

    # Hidden layer gradient
    error_hidden = d_output.dot(W2.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    # ---- WEIGHT UPDATE ----
    W2 += hidden_output.T.dot(d_output) * lr
    W1 += X.T.dot(d_hidden) * lr

    # Print loss every 1000 steps
    if epoch % 1000 == 0:
        loss = np.mean(np.abs(error))
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# -----------------------------
# 6. Results
# -----------------------------
print("\nOutputs after training:")
print(final_output)
```
