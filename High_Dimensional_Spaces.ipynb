```python
# -*- coding: utf-8 -*-
"""
8_4_High_Dimensional_Spaces.py
Completed version with English comments
"""

import numpy as np
import matplotlib.pyplot as plt
import scipy.special as sci

# ==========================================
# 1. DISTANCE RATIOS IN HIGH DIMENSIONS
# ==========================================

np.random.seed(0)
n_data = 1000

# Generate 1000 data points in 2D, 100D, and 1000D
x_2D = np.random.normal(size=(2, n_data))
x_100D = np.random.normal(size=(100, n_data))
x_1000D = np.random.normal(size=(1000, n_data))

def distance_ratio(x):
    """Compute ratio of largest to smallest pairwise Euclidean distances."""
    n_points = x.shape[1]
    dists = []
    for i in range(n_points):
        for j in range(i + 1, n_points):
            d = np.linalg.norm(x[:, i] - x[:, j])
            dists.append(d)
    smallest_dist = np.min(dists)
    largest_dist = np.max(dists)
    return largest_dist / smallest_dist

print('Ratio of largest to smallest distance 2D: %3.3f' % (distance_ratio(x_2D)))
print('Ratio of largest to smallest distance 100D: %3.3f' % (distance_ratio(x_100D)))
print('Ratio of largest to smallest distance 1000D: %3.3f' % (distance_ratio(x_1000D)))

# ==========================================
# 2. VOLUME OF A HYPERSPHERE
# ==========================================

def volume_of_hypersphere(diameter, dimensions):
    """Compute the volume of a hypersphere given its diameter and dimensionality."""
    pi = np.pi
    radius = diameter / 2.0
    volume = (pi ** (dimensions / 2.0)) * (radius ** dimensions) / sci.gamma((dimensions / 2.0) + 1)
    return volume

diameter = 1.0
for c_dim in range(1, 11):
    print("Volume of unit diameter hypersphere in %d dimensions is %3.6f" %
          (c_dim, volume_of_hypersphere(diameter, c_dim)))

# ==========================================
# 3. PROPORTION OF VOLUME IN OUTER 1%
# ==========================================

def get_prop_of_volume_in_outer_1_percent(dimension):
    """Compute proportion of hypersphere volume in the outer 1%% of the radius."""
    full_vol = volume_of_hypersphere(1.0, dimension)
    inner_vol = volume_of_hypersphere(0.99, dimension)
    proportion = 1 - (inner_vol / full_vol)
    return proportion

for c_dim in [1, 2, 10, 20, 50, 100, 150, 200, 250, 300]:
    print('Proportion of volume in outer 1 percent of radius in %d dimensions = %3.6f' %
          (c_dim, get_prop_of_volume_in_outer_1_percent(c_dim)))
```
